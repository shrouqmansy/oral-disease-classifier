```python
import os
import time
import json
import shutil
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.utils import image_dataset_from_directory
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.applications import MobileNetV2, ResNet50, EfficientNetB0
from tensorflow.keras.applications import mobilenet_v2, resnet50

from sklearn.metrics import classification_report, confusion_matrix
from tensorflow.keras.applications import DenseNet121, densenet
```

```python
!pip install -q kagglehub split-folders
import splitfolders
import kagglehub

# downloads the dataset and returns its path
download_path = kagglehub.dataset_download("salmansajid05/oral-diseases")
print(download_path)
```

```python
# find the folder that contains the disease folders
base_path = None
for root, dirs, files in os.walk(download_path):
    if "Gingivitis" in dirs:
        base_path = root
        break

clean_dir = "/content/clean_data"
dataset_dir = "/content/dataset"

batch_size = 32
image_width = 224
image_height = 224
epochs = 20

print(base_path)
```

```python
# the 3 augmented folders are used so class sizes are close
sources = {
    "Gingivitis": base_path + "/Gingivitis/Gingivitis",
    "hypodontia": base_path + "/hypodontia/hypodontia",
    "Calculus": base_path + "/Calculus/Calculus",
    "Caries": base_path + "/Data caries/Data caries/caries augmented data set/preview",
    "Mouth_Ulcer": base_path + "/Mouth Ulcer/Mouth Ulcer/Mouth_Ulcer_augmented_DataSet/preview",
    "Tooth_Discoloration": base_path + "/Tooth Discoloration/Tooth Discoloration /Tooth_discoloration_augmented_dataser/preview",
}

for class_name, folder in sources.items():
    os.makedirs(clean_dir + "/" + class_name, exist_ok=True)
    for file in os.listdir(folder):
        if file.lower().endswith((".jpg", ".jpeg", ".png")):
            shutil.copy(folder + "/" + file, clean_dir + "/" + class_name + "/" + file)

splitfolders.ratio(clean_dir, output=dataset_dir, seed=42, ratio=(0.7, 0.15, 0.15))
```

```python
train_ds = image_dataset_from_directory(
    dataset_dir + "/train",
    image_size=(image_height, image_width),
    batch_size=batch_size,
    seed=0
)

val_ds = image_dataset_from_directory(
    dataset_dir + "/val",
    image_size=(image_height, image_width),
    batch_size=batch_size
)

# no shuffle so predictions match the labels later
test_ds = image_dataset_from_directory(
    dataset_dir + "/test",
    image_size=(image_height, image_width),
    batch_size=batch_size,
    shuffle=False
)

class_names = train_ds.class_names
num_classes = len(class_names)
print(class_names)
print(num_classes)
```

```python
for class_name in class_names:
    count = len(os.listdir(dataset_dir + "/train/" + class_name))
    print(class_name, "->", count)

plt.figure(figsize=(10, 10))
for images, labels in train_ds.take(1):
    for i in range(9):
        plt.subplot(3, 3, i + 1)
        plt.imshow(images[i].numpy().astype("uint8"))
        plt.title(class_names[labels[i]])
        plt.axis("off")
plt.show()
```

```python
data_augmentation = tf.keras.Sequential([layers.RandomFlip("horizontal"),layers.RandomRotation(0.1),layers.RandomZoom(0.1),])
```

```python
def plot_history(history, title):
    plt.figure(figsize=(12, 4))

    plt.subplot(1, 2, 1)
    plt.plot(history.history["accuracy"], label="train")
    plt.plot(history.history["val_accuracy"], label="val")
    plt.title(title + " - Accuracy")
    plt.legend()

    plt.subplot(1, 2, 2)
    plt.plot(history.history["loss"], label="train")
    plt.plot(history.history["val_loss"], label="val")
    plt.title(title + " - Loss")
    plt.legend()
    plt.show()


# stop when val loss stops improving
early_stop = EarlyStopping(monitor="val_loss", patience=3, restore_best_weights=True)


def train_model(model, name):
    start = time.time()
    history = model.fit(train_ds, validation_data=val_ds, epochs=epochs, callbacks=[early_stop])
    train_time = time.time() - start
    plot_history(history, name)
    return history, train_time
```

```python
cnn_model = models.Sequential([
    layers.Input(shape=(image_height, image_width, 3)),
    data_augmentation,
    layers.Rescaling(1./255),

    layers.Conv2D(32, 3, activation="relu"),
    layers.BatchNormalization(),
    layers.MaxPooling2D(),

    layers.Conv2D(64, 3, activation="relu"),
    layers.BatchNormalization(),
    layers.MaxPooling2D(),

    layers.Conv2D(128, 3, activation="relu"),
    layers.BatchNormalization(),
    layers.MaxPooling2D(),

    layers.Flatten(),
    layers.Dense(128, activation="relu"),
    layers.Dropout(0.5),
    layers.Dense(num_classes, activation="softmax")
])

cnn_model.compile(optimizer=Adam(1e-3),
                  loss="sparse_categorical_crossentropy",
                  metrics=["accuracy"])

cnn_model.summary()
```

```python
history_cnn, time_cnn = train_model(cnn_model, "Custom CNN")
```

```python
# EfficientNet has its own preprocessing built in
pretrained = {"EfficientNetB0": (EfficientNetB0, None), "DenseNet121": (DenseNet121, densenet.preprocess_input)}

def build_pretrained(name, lr=1e-3, dropout=0.3):
    base_class, preprocess = pretrained[name]
    base_model = base_class(weights="imagenet", include_top=False, input_shape=(image_height, image_width, 3))
    base_model.trainable = False

    inputs = layers.Input(shape=(image_height, image_width, 3))
    x = data_augmentation(inputs)
    if preprocess is not None:
        x = preprocess(x)
    x = base_model(x, training=False)
    x = layers.GlobalAveragePooling2D()(x)
    x = layers.Dropout(dropout)(x)
    outputs = layers.Dense(num_classes, activation="softmax")(x)

    model = tf.keras.Model(inputs, outputs)
    model.compile(optimizer=Adam(lr), loss="sparse_categorical_crossentropy", metrics=["accuracy"])
    return model
```

```python
efficientnet = build_pretrained("EfficientNetB0")
history_eff, time_eff = train_model(efficientnet, "EfficientNetB0")

densenet_model = build_pretrained("DenseNet121")
history_dense, time_dense = train_model(densenet_model, "DenseNet121")
```

```python
# change to the better model after seeing cell 12 results
tune_name = "EfficientNetB0"
results = []

for lr in [1e-2, 1e-3, 1e-4]:
    for dropout in [0.2, 0.5]:
        model = build_pretrained(tune_name, lr=lr, dropout=dropout)
        h = model.fit(train_ds, validation_data=val_ds, epochs=5, verbose=0)
        val_acc = max(h.history["val_accuracy"])
        results.append((lr, dropout, val_acc))
        print("lr =", lr, "| dropout =", dropout, "| val_acc =", round(val_acc, 4))

best_lr, best_dropout, best_acc = max(results, key=lambda r: r[2])
print("Best -> lr:", best_lr, "dropout:", best_dropout, "val_acc:", round(best_acc, 4))
```

```python
tuned_model = build_pretrained(tune_name, lr=best_lr, dropout=best_dropout)
history_tuned, time_tuned = train_model(tuned_model, tune_name + " (tuned)")
```

```python
all_models = {"Custom CNN": (cnn_model, time_cnn), "EfficientNetB0": (efficientnet, time_eff), "DenseNet121": (densenet_model, time_dense), "EfficientNetB0 (tuned)": (tuned_model, time_tuned), "EfficientNetB0 (fine-tuned)": (efficientnet_ft, time_eff + time_ft)}

rows = []
for name, (model, t) in all_models.items():
    loss, acc = model.evaluate(test_ds, verbose=0)
    rows.append({"Model": name, "Test Accuracy": round(acc, 4), "Test Loss": round(loss, 4), "Parameters (M)": round(model.count_params() / 1e6, 2), "Train Time (min)": round(t / 60, 1)})

results_df = pd.DataFrame(rows)
print(results_df)
```

```python
best_name = results_df.sort_values("Test Accuracy", ascending=False).iloc[0]["Model"]
best_model = all_models[best_name][0]
print("Best model:", best_name)

y_true = np.concatenate([y for x, y in test_ds])
y_pred = np.argmax(best_model.predict(test_ds), axis=1)
print(classification_report(y_true, y_pred, target_names=class_names))

cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=class_names, yticklabels=class_names)
plt.xlabel("Predicted")
plt.ylabel("True")
plt.show()
```
